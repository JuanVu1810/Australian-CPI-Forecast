## 15. Documented limitations and follow-ups

[![content: AI-generated](../_static/badges/ai-generated.svg)](ai-acknowledgement) [![content: human-reviewed & edited](../_static/badges/human-reviewed-edited.svg)](ai-acknowledgement)

- **Trimmed mean is its own target, not a regressor.** Headline `cpi_yoy` is stationary (level ADF p=0.0299), while `trimmed_mean_cpi_yoy` isn't at level (ADF p=0.1896) but is after first differencing. Lagged trimmed mean does lead headline's quarterly change (lag-1 correlation 0.34 and Granger p=0.004 for its own quarterly change, 0.26 and p=0.055 for the YoY series) and its VIF is fine (3.427), so it could work as a forecast-safe regressor if release timing were handled. Because it's another CPI measure, not an outside indicator, we keep it out of the macro pool and model it as a separate target. The headline-minus-trimmed-mean gap shows no signal (best correlation 0.17, Granger p=0.358).
- **Trimmed mean needs its own feature screen, but the two screens agree more than we first thought.** Section 11 reruns the screen with trimmed mean as the target. Business inflation expectations lead both targets (headline r=0.46, Granger p below 0.0001; trimmed mean r=0.68, p below 0.0001). The cash rate is a clear signal for headline (r=0.37, p=0.0007) but not for trimmed mean (r=0.29, p=0.71), and PPI is weaker for trimmed mean (p=0.18, against 0.04 for headline). An earlier version of this screen on the YoY series found commodity and oil terms leading trimmed mean. On quarterly changes they show no Granger signal (p=0.17 to 0.27). Rolling-correlation stability (Section 12) is headline-only.
- **The screens are descriptive, and every BIC lag is 1.** They use the full 1995-2025 sample, which includes the years the walk-forward evaluation tests (it starts after 40 quarters of training data). The shorter-history predictors can't be screened on that initial window alone (household spending has 49 observations in total), so we didn't restrict the screens to training data. BIC picks a one-quarter lag for all 22 series, so they say little about longer leads. The Elastic Net feature sets in [§4.2](../04_modeling/42_elastic_net) were chosen from an earlier screen on the YoY series and haven't been rebuilt from this one. What justifies them is the walk-forward RMSE, not the screen.
- **Seasonality tells us little here.** The STL decomposition runs on the growth rates, not a CPI level index. Headline `cpi_yoy` has seasonal strength 0.000 and trend strength 0.842, and trimmed mean has 0.000 and 0.875. Both are trend-dominated, which says nothing about seasonality in a level index.
- **Release dates are assumed.** `data/metadata/series_availability.csv` is an assumption-backed lag table, not an official ABS or RBA release calendar, and the curated dataset uses revised history, not real-time vintages. Before any live forecast-origin decision, replace it with observation-level release dates and an `available_at_forecast_origin` field.

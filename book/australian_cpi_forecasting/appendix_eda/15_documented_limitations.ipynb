## 15. Documented limitations and follow-ups

[![content: AI-generated](../_static/badges/ai-generated.svg)](ai-acknowledgement) [![content: human-reviewed & edited](../_static/badges/human-reviewed-edited.svg)](ai-acknowledgement)

- **Trimmed mean is its own target, not a regressor.** Headline `cpi_yoy` is stationary (level ADF p=0.0299), while `trimmed_mean_cpi_yoy` isn't at level (ADF p=0.1896) but is after first differencing. Lagged trimmed mean does lead headline (best lag-2 correlation 0.4773, Granger p=0.0234) and its VIF is fine (3.427), so it could work as a forecast-safe regressor if release timing were handled. Because it's another CPI measure, not an outside indicator, we keep it out of the macro pool and model it as a separate target. The headline-minus-trimmed-mean gap is mixed: a high lag-1 correlation (0.6592), but Granger p=0.8141 at that lag.
- **Trimmed mean needs its own feature screen.** Section 11 reruns the screen with trimmed mean as the target, and the candidates change. Headline's strongest signals are business inflation expectations (correlation 0.7407, Granger p=0.0000), PPI (0.5556, p=0.0066) and the cash rate (0.5112). For trimmed mean those weaken sharply: expectations 0.2803 (p=0.4600), PPI 0.2459 (p=0.6868), cash rate 0.1198 (p=0.6177). Trimmed mean is led instead by lagged commodity growth (0.3226, p=0.0360) and WTI oil growth (0.3039, p=0.0330), with Brent as a shorter-sample alternative (0.3088, p=0.0372). So its Elastic Net starts from a smaller commodity and oil specification, not the headline set. Rolling-correlation stability (Section 12) is headline-only.
- **Seasonality tells us little here.** The STL decomposition runs on the growth rates, not a CPI level index. Headline `cpi_yoy` has seasonal strength 0.000 and trend strength 0.842, and trimmed mean has 0.000 and 0.875. Both are trend-dominated, which says nothing about seasonality in a level index.
- **Release dates are assumed.** `data/metadata/series_availability.csv` is an assumption-backed lag table, not an official ABS or RBA release calendar, and the curated dataset uses revised history, not real-time vintages. Before any live forecast-origin decision, replace it with observation-level release dates and an `available_at_forecast_origin` field.
